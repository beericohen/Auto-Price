```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df = pd.read_csv('./Car_Data_Clean.csv')
```

```python

print(df.describe())
```

```python
#price distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['price'], bins=30, kde=True)
plt.title('Price Distribution')
plt.xlabel('Price (ILS)')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('price_distribution.png')
plt.show()
```

```python
#average cost for manufacturer
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='manufacturer', y='price')
plt.title('Price by Manufacturer')
plt.xlabel('Manufacturer')
plt.ylabel('Price (ILS)')
plt.tight_layout()
plt.savefig('price_by_manufacturer.png')
plt.show()
```

```python
#mileage vs price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='mileage', y='price', hue='manufacturer')
plt.title('Price vs Mileage')
plt.xlabel('Mileage (KM)')
plt.ylabel('Price (ILS)')
plt.tight_layout()
plt.savefig('price_vs_mileage.png')
plt.show()
```

```python
#year vs price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='year', y='price', hue='manufacturer')
plt.title('Price vs Year')
plt.xlabel('Year')
plt.ylabel('Price (ILS)')
plt.tight_layout()
plt.savefig('price_vs_year.png')
plt.show()
```

```python
#how fuel type affects price
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='fuel', y='price')
plt.title('Price by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Price (ILS)')
plt.tight_layout()
plt.savefig('price_by_fuel.png')
plt.show()
```

```python
#heatmap
plt.figure(figsize=(8, 6))
corr = df[['year', 'mileage', 'price', 'hand']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('correlation_heatmap.png')
plt.show()
```

```python
#average price by model
plt.figure(figsize=(14, 6))
model_avg = df.groupby('model')['price'].median().sort_values(ascending=False)
sns.barplot(x=model_avg.index, y=model_avg.values)
plt.title('Median Price by Model')
plt.xlabel('Model')
plt.ylabel('Median Price (ILS)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('price_by_model.png')
plt.show()
```
