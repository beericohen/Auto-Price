```python
import pandas as pd
import numpy as np
import joblib
```

```python
df = pd.read_csv(r'C:\Users\USER\Documents\Projects\AutoPrice\Auto-Price\data/autoboom_clean.csv', index_col=False)
```

```python
df = df.fillna(0)
```

```python
dummies = pd.get_dummies(df, dtype=int)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
Yscaler = MinMaxScaler()
```

```python
df.columns
```

```python
cols_to_scale = ['year', 'hand', 'engine_liters', 'horsepower', 'mileage']
dummies[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])
dummies['price'] = Yscaler.fit_transform(df[['price']])
```

```python
joblib.dump(Yscaler, r'C:\Users\USER\Documents\Projects\AutoPrice\Auto-Price\data/minmax_scaler.pkl')
joblib.dump(scaler, r'C:\Users\USER\Documents\Projects\AutoPrice\Auto-Price\data/scaler.pkl')
```

```python
dummies.to_csv(r'C:\Users\USER\Documents\Projects\AutoPrice\Auto-Price\data/preprocessing.csv', index=False)
```
